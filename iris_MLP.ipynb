```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/iris.csv")
```

```python
df.head()
```

```python
x= df.drop("Name", axis = 1)
x
```

```python
y= pd.get_dummies(df.Name)
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3) #30% of the data is used for testing
```

```python
x_train.head()
```

```python
y_train
```

```python
import tensorflow as tf
import keras
from keras import layers
```

```python
model = keras.Sequential()
model.add(layers.Dense(6, activation="sigmoid"))
model.add(layers.Dense(3, activation = 'softmax'))
```

```python
model.compile(optimizer='Adam', loss= 'categorical_crossentropy', metrics=["accuracy"])
```

```python
model.fit(x_train, y_train, epochs= 50, batch_size= 5)
```

```python
score = model.evaluate(x_test, y_test)
score
```

```python
#An accuracy of 0.6039 (60.39%) is achieved on the test set using untuned model.
#The model is trained for 50 epochs with a batch size of 5.
```

```python
#Code for hyperparameter tuning using Keras Tuner
```

```python
def model_builder(hp):
  model = tf.keras.Sequential()
  model.add(tf.keras.Input(shape=(4,)))

  hp_activation = hp.Choice('activation', values=['relu', 'sigmoid', 'tanh', 'softmax' ])
  hp_layer_1 = hp.Int('layer_1', min_value=1, max_value=10, step=1)
  hp_learning_rate = hp.Choice('learning_rate', values=[1e-2, 1e-3, 1e-4])

  model.add(tf.keras.layers.Dense(units=hp_layer_1, activation=hp_activation))
  model.add(tf.keras.layers.Dense(3, activation='softmax'))

  model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=hp_learning_rate),
                loss=tf.keras.losses.CategoricalCrossentropy(),
                metrics=['accuracy'])
  
  return model
```

```python
import keras_tuner as kt

tuner = kt.Hyperband(model_builder,
                     objective='val_accuracy',
                     max_epochs=10,
                     factor=3,
                     directory='dir',
                     project_name='x')
```

```python
stop_early = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5)
```

```python
tuner.search(x_train, y_train, epochs=50, validation_split=0.2, callbacks=[stop_early])
```

```python
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
```

```python
print(f"""
The hyperparameter search is complete. The optimal number of units in the first hidden layer is {best_hps.get('layer_1')} and the optimal learning rate for the optimizer is {best_hps.get('learning_rate')}.
The optimal activation function is {best_hps.get('activation')}.""")
```

```python
model_tuned = tuner.hypermodel.build(best_hps)
history = model_tuned.fit(x_train, y_train, epochs=50, batch_size=5)
```

```python
score = model_tuned.evaluate(x_test, y_test)
score
```

```python
#An accuracy of 0.9243 (92.43%) is achieved on the test set using the hyper-parameter tuned model.
```
